###### Weekly time-series plot

```python
path = 'C:/Users/Asifuzzaman.jishan/Downloads/ENTSO-E_Data_analysis_and_ml/Graph_timeseries_weekly_biomass_and_fossil brown coal'
os.mkdir(path)
```

```python
# Define the start and end dates for slicing the data
start_date = df.index.min()
end_date = df.index.max()

# Initialize the start date for the first plot
current_start = start_date

while current_start < end_date:
    current_end = current_start + pd.Timedelta(days=7)
    weekly_data = df[current_start:current_end]

    plt.figure(figsize=(14, 7))
    plt.plot(weekly_data.index, weekly_data['Biomass'], label='Biomass', color='blue')
    plt.plot(weekly_data.index, weekly_data['Fossil Brown coal/Lignite'], label='Fossil Brown coal/Lignite', color='orange')

    # Setting the x-axis to display dates in a more readable format
    plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=3))  # Major ticks every 6 hours
    plt.gca().xaxis.set_minor_locator(mdates.MinuteLocator(interval=15))  # Minor ticks every 15 minutes
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d-%m %H:%M'))

    plt.gcf().autofmt_xdate(rotation=90)  # Rotate date labels to prevent overlap

    plt.title(f'Time Series Plot: {current_start.date()} to {current_end.date()}')
    plt.xlabel('Date')
    plt.ylabel('Values')
    plt.legend()
    plt.savefig(f'C:/Users/Asifuzzaman.jishan/Downloads/ENTSO-E_Data_analysis_and_ml/Graph_timeseries_weekly_biomass_and_fossil brown coal/weekly_timeseries_{current_start.date()}_to_{current_end.date()}.png')
    plt.show()

    # Move to the next 7-day period
    current_start = current_end
```

#### Mean, Median, Quantile, Percentile (Biomass and Fossil Brown coal/Lignite)

```python
# Define the start and end dates for slicing the data
start_date = df.index.min()
end_date = df.index.max()

# Initialize the start date for the first plot
current_start = start_date

# Lists to store weekly data
biomass_data = []
fossil_data = []

while current_start < end_date:
    current_end = current_start + pd.Timedelta(days=7)
    weekly_data = df[current_start:current_end]

    biomass_data.append(weekly_data['Biomass'])
    fossil_data.append(weekly_data['Fossil Brown coal/Lignite'])

    current_start = current_end

# Concatenate all weekly data into single series for quantile calculation
biomass_df = pd.concat(biomass_data, axis=1)
fossil_df = pd.concat(fossil_data, axis=1)

# Calculate quantiles
biomass_quantiles = biomass_df.quantile([0.25, 0.5, 0.75], axis=1)
fossil_quantiles = fossil_df.quantile([0.25, 0.5, 0.75], axis=1)

# Calculate mean and median
biomass_mean = biomass_df.mean(axis=1)
biomass_median = biomass_df.median(axis=1)
fossil_mean = fossil_df.mean(axis=1)
fossil_median = fossil_df.median(axis=1)

# Sort data by index for plotting
biomass_quantiles = biomass_quantiles.sort_index()
fossil_quantiles = fossil_quantiles.sort_index()
biomass_mean = biomass_mean.sort_index()
biomass_median = biomass_median.sort_index()
fossil_mean = fossil_mean.sort_index()
fossil_median = fossil_median.sort_index()

fig, axs = plt.subplots(2, 1, figsize=(18, 14), sharex=True)  # Increase figure size

# Plot Biomass quantiles, mean, and median
axs[0].fill_between(biomass_df.index, biomass_quantiles.loc[0.25], biomass_quantiles.loc[0.75], color='blue', alpha=0.5, label='Biomass 25th-75th Percentile')
axs[0].plot(biomass_df.index, biomass_quantiles.loc[0.5], color='navy', linestyle='-', label='Biomass Median')
axs[0].plot(biomass_df.index, biomass_quantiles.loc[0.25], color='lightblue', linestyle='--', label='Biomass 25th Percentile')
axs[0].plot(biomass_df.index, biomass_quantiles.loc[0.75], color='darkblue', linestyle='-.', label='Biomass 75th Percentile')
axs[0].plot(biomass_df.index, biomass_mean, color='red', linestyle=':', label='Biomass Mean')
axs[0].set_title('Biomass with Quantiles, Mean, and Median')
axs[0].set_ylabel('Values')
axs[0].legend()

# Plot Fossil Brown coal/Lignite quantiles, mean, and median
axs[1].fill_between(fossil_df.index, fossil_quantiles.loc[0.25], fossil_quantiles.loc[0.75], color='orange', alpha=0.5, label='Fossil 25th-75th Percentile')
axs[1].plot(fossil_df.index, fossil_quantiles.loc[0.5], color='darkorange', linestyle='-', label='Fossil Median')
axs[1].plot(fossil_df.index, fossil_quantiles.loc[0.25], color='yellow', linestyle='--', label='Fossil 25th Percentile')
axs[1].plot(fossil_df.index, fossil_quantiles.loc[0.75], color='orangered', linestyle='-.', label='Fossil 75th Percentile')
axs[1].plot(fossil_df.index, fossil_mean, color='red', linestyle=':', label='Fossil Mean')
axs[1].set_title('Fossil Brown coal/Lignite with Quantiles, Mean, and Median')
axs[1].set_xlabel('Date')
axs[1].set_ylabel('Values')
axs[1].legend()

# Setting the x-axis to display dates in a more readable format
for ax in axs:
    ax.xaxis.set_major_locator(mdates.MonthLocator())
    ax.xaxis.set_minor_locator(mdates.WeekdayLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))
    ax.tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.savefig('C:/Users/Asifuzzaman.jishan/Downloads/ENTSO-E_Data_analysis_and_ml/Graph_timeseries_Mean, Median, Quantile, Percentile/Biomass_fossil.png')
plt.show()
```

#### Clustering 

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# Aggregate data into 7-day windows (672 measurements)
def create_rolling_windows(data, window_size):
    return np.array([data[i:i + window_size].values.flatten() for i in range(0, len(data) - window_size + 1, window_size)])
```

```python
# Aggregate data into 7-day windows
window_size = 4 * 24 * 7  # 4 measurements per hour, 24 hours, 7 days, Total 672 dimen. 
rolling_windows = create_rolling_windows(df, window_size)
```

```python
# Create a DataFrame from the rolling windows
columns = [f'{col}_{i}' for i in range(window_size) for col in df.columns]
windows_df = pd.DataFrame(rolling_windows, columns=columns)
```

```python
# Normalize the features using MinMaxScaler
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(windows_df)
```

```python
# Apply PCA for dimensionality reduction
pca = PCA(n_components=12)  # Adjusted to retain around 95% variance
principal_components = pca.fit_transform(scaled_data)
```

```python
# Create a DataFrame with the principal components
pca_df = pd.DataFrame(data=principal_components, columns=[f'Principal Component {i+1}' for i in range(12)])
```

```python
# Elbow Method to find the optimal number of clusters
inertia = []
silhouette_scores = []
K = range(2, 11)  # You can adjust the range

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(pca_df)
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(pca_df, kmeans.labels_))
```

```python
# Plot Inertia (Elbow Method)
plt.figure(figsize=(8, 5))
plt.plot(K, inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal Number of Clusters')
plt.show()
```

```python
# Plot Silhouette Scores
plt.figure(figsize=(8, 5))
plt.plot(K, silhouette_scores, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Scores for Optimal Number of Clusters')
plt.show()
```

```python
# Print all scores
print("Inertia for different number of clusters:")
for k, score in zip(K, inertia):
    print(f"Number of clusters: {k}, Inertia: {score}")
```

```python
print("\nSilhouette Scores for different number of clusters:")
for k, score in zip(K, silhouette_scores):
    print(f"Number of clusters: {k}, Silhouette Score: {score}")
```

```python
# Choose the best number of clusters based on Silhouette Score
best_n_clusters = K[np.argmax(silhouette_scores)]
print(f'Best number of clusters based on Silhouette Score: {best_n_clusters}')
```

```python
# Apply k-means clustering with the optimal number of clusters
kmeans = KMeans(n_clusters=best_n_clusters, random_state=42)
kmeans_labels = kmeans.fit_predict(pca_df)
windows_df['Cluster_KMeans'] = kmeans_labels
```

```python
# Evaluate the clustering performance using silhouette score
kmeans_silhouette = silhouette_score(pca_df, kmeans_labels)
print(f"Silhouette Score for K-means with {best_n_clusters} clusters: {kmeans_silhouette}")
```

```python
# Visualize the clusters
plt.figure(figsize=(8, 5))
plt.scatter(pca_df.iloc[:, 0], pca_df.iloc[:, 1], c=kmeans_labels, cmap='viridis')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.title(f'K-means Clustering with {best_n_clusters} Clusters')
plt.colorbar(label='Cluster')
plt.show()
```

The primary difference between the two codes is the explicit setting of the n_init parameter in the second code, which ensures consistent and stable clustering results. 

The n_init parameter in the KMeans clustering algorithm specifies the number of times the algorithm will run with different centroid seeds. The final results will be the best output of n_init consecutive runs in terms of inertia, which is a measure of how internally coherent clusters are.

Understanding n_init
Default Value: The default value of n_init is 10 in versions of scikit-learn prior to 1.4. This means the algorithm runs 10 times with different initializations and returns the best solution.
Stability: A higher value for n_init increases the likelihood of finding a global optimum, as the algorithm runs multiple times with different initial centroids.
Consistency: Setting n_init explicitly to a consistent value (e.g., 10) ensures that the clustering results are stable across different runs, given the same random state.

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Function to create rolling windows and retain the start index of each window
def create_rolling_windows_with_indices(data, window_size):
    windows = []
    indices = []
    for i in range(0, len(data) - window_size + 1, window_size):
        windows.append(data[i:i + window_size].values.flatten())
        indices.append(data.index[i])  # store the starting index of each window
    return np.array(windows), indices

# Define the window size
window_size = 4 * 24 * 7  # 4 measurements per hour, 24 hours, 7 days, Total 672 dimensions

# Aggregate data into 7-day windows and retain indices
rolling_windows, window_indices = create_rolling_windows_with_indices(df, window_size)

# Create a DataFrame from the rolling windows
columns = [f'{col}_{i}' for i in range(window_size) for col in df.columns]
windows_df = pd.DataFrame(rolling_windows, columns=columns)

# Normalize the features using MinMaxScaler
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(windows_df)

# Apply PCA for dimensionality reduction
pca = PCA(n_components=12)  # Adjusted to retain around 95% variance
principal_components = pca.fit_transform(scaled_data)

# Create a DataFrame with the principal components
pca_df = pd.DataFrame(data=principal_components, columns=[f'Principal Component {i+1}' for i in range(12)])

# Elbow Method to find the optimal number of clusters
inertia = []
silhouette_scores = []
K = range(2, 11)  # You can adjust the range

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(pca_df)
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(pca_df, kmeans.labels_))

# Plot Inertia (Elbow Method)
plt.figure(figsize=(8, 5))
plt.plot(K, inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal Number of Clusters')
plt.show()

# Plot Silhouette Scores
plt.figure(figsize=(8, 5))
plt.plot(K, silhouette_scores, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Scores for Optimal Number of Clusters')
plt.show()

# Print all scores
print("Inertia for different number of clusters:")
for k, score in zip(K, inertia):
    print(f"Number of clusters: {k}, Inertia: {score}")

print("\nSilhouette Scores for different number of clusters:")
for k, score in zip(K, silhouette_scores):
    print(f"Number of clusters: {k}, Silhouette Score: {score}")

# Choose the best number of clusters based on Silhouette Score
best_n_clusters = K[np.argmax(silhouette_scores)]
print(f'Best number of clusters based on Silhouette Score: {best_n_clusters}')

# Apply k-means clustering with the optimal number of clusters
kmeans = KMeans(n_clusters=best_n_clusters, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(pca_df)
windows_df['Cluster_KMeans'] = kmeans_labels

# Evaluate the clustering performance using silhouette score
kmeans_silhouette = silhouette_score(pca_df, kmeans_labels)
print(f"Silhouette Score for K-means with {best_n_clusters} clusters: {kmeans_silhouette}")

# Visualize the clusters
plt.figure(figsize=(8, 5))
plt.scatter(pca_df.iloc[:, 0], pca_df.iloc[:, 1], c=kmeans_labels, cmap='viridis')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.title(f'K-means Clustering with {best_n_clusters} Clusters')
plt.colorbar(label='Cluster')
plt.show()
```
